```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
pi=np.pi
```

```python
np.random.rand(5)/10
```

```python
f1=8.5
f2=7

sampling=100
duration=10
N=sampling*duration

x=np.linspace(0,duration,N,endpoint=False)

y1=pi*np.sin(2*pi*f1*x+(pi/3))
y2=7*np.sin(2*pi*f2*x)
y=y1+y2
plt.plot(x,y)
```

```python
raw_fft = np.fft.fft(y)
len_raw_fft=len(raw_fft)
raw_fft = raw_fft[:len_raw_fft//2]

fft_abs = np.abs(raw_fft)
fft = fft_abs.tolist()

freq=np.fft.fftfreq(N,1/sampling)
len_freq=len(freq)
freq=freq[:len_freq//2]

plt.plot(freq,fft)
```

```python
sort_fft=sorted(fft,reverse=True)
sort_fft
```

```python
def arctan(r,i):
    if i==0:
        return pi/2
    else:
        return float(np.arctan((r)/(i)))

d={}
for i in sort_fft:
    c=raw_fft[fft.index(i)]
    f=float(freq[fft.index(i)])
    phase=arctan((c.real),(c.imag))
    amp=i/(duration*50)
    d[f]=[amp,phase]

print(d)
```
