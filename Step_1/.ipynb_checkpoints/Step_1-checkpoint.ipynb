```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
pi=np.pi
```

```python
f1=50
f2=60
f3=132
f4=2

sampling=200
duration=10
N=sampling*duration
x=np.linspace(0,duration,N,endpoint=False)
y1=-0.5*np.sin(2*pi*f1*x)
y2=-0.75*np.sin(2*pi*f2*x)
y3=np.sin(2*pi*f3*x)
y4=np.sin(2*pi*f4*x)
y=y1+y2+y3+y4

plt.plot(x,y)
```

```python
raw_fft = np.fft.fft(y)
angle=np.angle(raw_fft)
print(angle)
fft_abs = np.abs(raw_fft)
fft = fft_abs.tolist()

freq=np.fft.fftfreq(N,1/sampling)
print(len(freq))
plt.plot(angle,fft)
```
